(B) Comparative Analysis

1. Supported Data Mining Tasks

All three datasets directly support core course techniques such as frequent itemset mining and association rule learning, since each can be structured into transaction baskets. Instacart and Online Retail II are particularly suitable for Apriori or FP-Growth due to clear basket identifiers (order_id and InvoiceNo). Complete Journey additionally supports time-indexed analysis because transactions span multiple weeks and campaigns. The main distinction lies in beyond-course potential. Instacart is strongest for sequential pattern mining and next-basket recommendation due to ordered user purchase histories. Online Retail II introduces a text modality via product descriptions, enabling transformer-based embeddings and high-utility itemset mining. Complete Journey provides the causal structure, allowing uplift modeling and campaign-effect estimation through coupon and promotion joins.

2. Data Quality Issues

Each dataset presents realistic but different data challenges. Instacart contains missing `days_since_prior_order` values for first purchases and lacks price information, limiting economic analysis. Online Retail II has more traditional noise, including missing `CustomerID`, negative quantities representing returns, cancellation invoices (InvoiceNo beginning with C), and inconsistent product descriptions. Complete Journey is structurally complex rather than noisy. Demographic data is incomplete for some households, and multi-table joins introduce integration challenges. Overall, Online Retail II requires the most cleaning, Instacart has feature limitations, and Complete Journey requires careful relational handling.

3. Algorithmic Feasibility

Online Retail II is the most computationally manageable, with 1.07M rows in a single table, making FP-Growth, clustering, and text embeddings feasible on a standard laptop. Complete Journey (1.47M transactions) remains manageable, though incorporating the 20M+ promotions table increases memory requirements. Instacart is the largest (3.4M orders, 30M+ line items), and naive Apriori on the full product set may be infeasible without pruning. Instacart is computationally heavier but still feasible with appropriate filtering strategies.

4. Bias Considerations

All three datasets reflect sampling and behavioral biases. Instacart captures only users of a specific online grocery platform, limiting generalizability to broader retail contexts. Online Retail II reflects transactions from a single UK retailer with a substantial wholesale component. Complete Journey represents a panel from a particular retail environment, and demographic attributes may not represent the full population. Additionally, retail datasets inherently reflect popularity bias, where frequently purchased items become increasingly dominant in patterns. Therefore, findings should be interpreted within platform and sampling constraints.

```python
# ===========================
#       Instacart EDA
# ===========================
# Assumptions:
# - Folder already contains: orders.csv, order_products__prior.csv, order_products__train.csv,
#   products.csv, aisles.csv, departments.csv

import os
import gc
import numpy as np
import pandas as pd

import matplotlib
matplotlib.rcParams["figure.dpi"] = 120
matplotlib.rcParams["figure.figsize"] = (5, 4)
import matplotlib.pyplot as plt

import seaborn as sns
sns.set_style("darkgrid")
color = sns.color_palette()

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 140)

# Config
DATA_DIR = "./instacart"   # <-- change to your folder path
SAMPLE_USERS = None        # e.g., 50000 to sample users for faster EDA, or None for full
LIMIT_PRIOR_ROWS = None    # e.g., 5_000_000 to limit prior rows for speed, or None for full
TOP_N_AISLES = 25
TOP_N_DEPARTMENTS = 21
RANDOM_SEED = 42

def _path(name):
    p = os.path.join(DATA_DIR, name)
    if not os.path.exists(p):
        raise FileNotFoundError(f"Missing file: {p}. Update DATA_DIR to the folder containing Instacart CSVs.")
    return p

# Helper overview function
def df_overview(df, name):
    print(f"\n=== {name} overview ===")
    print(df.head(3))
    print("\nDtypes:")
    print(df.dtypes)
    print("\nMissing values (count, %):")
    miss = df.isna().sum().sort_values(ascending=False)
    miss_pct = (miss / len(df) * 100).round(3)
    out = pd.DataFrame({"missing_count": miss, "missing_pct": miss_pct})
    print(out[out["missing_count"] > 0].head(20) if (miss > 0).any() else "No missing values.")

# 1) Load data
orders_dtypes = {
    "order_id": "int32",
    "user_id": "int32",
    "eval_set": "category",
    "order_number": "int16",
    "order_dow": "int8",
    "order_hour_of_day": "int8",
    "days_since_prior_order": "float32",
}
op_dtypes = {
    "order_id": "int32",
    "product_id": "int32",
    "add_to_cart_order": "int16",
    "reordered": "int8",
}
products_dtypes = {
    "product_id": "int32",
    "product_name": "string",
    "aisle_id": "int16",
    "department_id": "int16",
}
aisles_dtypes = {"aisle_id": "int16", "aisle": "string"}
depts_dtypes = {"department_id": "int16", "department": "string"}

print("Loading CSVs...")
orders = pd.read_csv(_path("orders.csv"), dtype=orders_dtypes)
products = pd.read_csv(_path("products.csv"), dtype=products_dtypes)
aisles = pd.read_csv(_path("aisles.csv"), dtype=aisles_dtypes)
departments = pd.read_csv(_path("departments.csv"), dtype=depts_dtypes)

op_prior = pd.read_csv(_path("order_products__prior.csv"), dtype=op_dtypes, nrows=LIMIT_PRIOR_ROWS)
op_train = pd.read_csv(_path("order_products__train.csv"), dtype=op_dtypes)

print("Shapes")
print("orders:", orders.shape)
print("op_prior:", op_prior.shape, "(limited)" if LIMIT_PRIOR_ROWS else "")
print("op_train:", op_train.shape)
print("products:", products.shape)
print("aisles:", aisles.shape)
print("departments:", departments.shape)

if SAMPLE_USERS is not None:
    rng = np.random.default_rng(RANDOM_SEED)
    sampled_users = rng.choice(
        orders["user_id"].unique(),
        size=min(SAMPLE_USERS, orders["user_id"].nunique()),
        replace=False
    )
    orders = orders[orders["user_id"].isin(sampled_users)].copy()

    sampled_order_ids = set(orders["order_id"].astype(int).tolist())
    op_prior = op_prior[op_prior["order_id"].isin(sampled_order_ids)].copy()
    op_train = op_train[op_train["order_id"].isin(sampled_order_ids)].copy()

    print(f"\nSampled users: {len(sampled_users)}")
    print("Filtered shapes -> orders:", orders.shape, "op_prior:", op_prior.shape, "op_train:", op_train.shape)

# Join tables
prod_meta = products.merge(aisles, on="aisle_id", how="left").merge(departments, on="department_id", how="left")

op_prior_m = op_prior.merge(prod_meta, on="product_id", how="left")
op_train_m = op_train.merge(prod_meta, on="product_id", how="left")

print("\nJoined prior/train with product metadata.")
print("op_prior_m:", op_prior_m.shape, "op_train_m:", op_train_m.shape)

# Tests
df_overview(orders, "orders")
df_overview(op_prior, "order_products__prior")
df_overview(op_train, "order_products__train")
df_overview(products, "products")

print("\n=== Sanity checks ===")
print("Unique order_id in orders:", orders["order_id"].nunique())
print("Unique user_id:", orders["user_id"].nunique())
print("Eval_set counts:\n", orders["eval_set"].value_counts(dropna=False))
print("days_since_prior_order missing should roughly match first orders:",
      orders["days_since_prior_order"].isna().mean().round(4))
```

```python
# Basket-level features
basket_size = op_prior_m.groupby("order_id", as_index=False).agg(
    basket_size=("product_id", "count"),
    unique_products=("product_id", "nunique"),
    reordered_items=("reordered", "sum"),
)
basket_size["reorder_ratio_in_basket"] = (basket_size["reordered_items"] / basket_size["basket_size"]).astype("float32")

orders_m = orders.merge(basket_size, on="order_id", how="left")
orders_m["basket_size"] = orders_m["basket_size"].fillna(0).astype("int32")
orders_m["unique_products"] = orders_m["unique_products"].fillna(0).astype("int32")
orders_m["reordered_items"] = orders_m["reordered_items"].fillna(0).astype("int32")
orders_m["reorder_ratio_in_basket"] = orders_m["reorder_ratio_in_basket"].fillna(0).astype("float32")

print("\n=== Basket size summary (prior orders only contribute) ===")
print(orders_m[["basket_size", "unique_products", "reordered_items", "reorder_ratio_in_basket"]].describe().T)

plt.figure()
orders_m.loc[orders_m["basket_size"] > 0, "basket_size"].clip(upper=100).hist(bins=50)
plt.title("Basket Size Distribution (clipped at 100)")
plt.xlabel("Items per order")
plt.ylabel("Count")
plt.show()

# Tests
# Basket size must equal product count per order
check_counts = (
    op_prior_m.groupby("order_id")["product_id"]
    .count()
    .reset_index(name="expected_count")
)

merged_check = orders_m.merge(check_counts, on="order_id", how="left")
mask = merged_check["expected_count"].notna()

assert (merged_check.loc[mask, "basket_size"] ==
        merged_check.loc[mask, "expected_count"]).all(), \
        "ERROR: Basket size does not match grouped product count!"

# Reorder ratio must be between 0 and 1
assert ((orders_m["reorder_ratio_in_basket"] >= 0) &
        (orders_m["reorder_ratio_in_basket"] <= 1)).all(), \
        "ERROR: Reorder ratio outside valid range [0,1]!"

# Reordered items cannot exceed basket size
assert (orders_m["reordered_items"] <= orders_m["basket_size"]).all(), \
        "ERROR: Reordered items exceed basket size!"
```

```python
# Basket–Product Matrix Sparsity

# Total baskets (prior orders only)
num_baskets = op_prior_m["order_id"].nunique()

# Total unique products (prior)
num_products = op_prior_m["product_id"].nunique()

# Total non-zero entries (each row in op_prior_m is one basket-product interaction)
non_zero_entries = len(op_prior_m)

# Total possible entries in basket-product matrix
total_possible_entries = num_baskets * num_products

# Sparsity calculation
sparsity = 1 - (non_zero_entries / total_possible_entries)

print("=== Basket-Product Matrix Sparsity ===")
print(f"Number of baskets: {num_baskets:,}")
print(f"Number of unique products: {num_products:,}")
print(f"Non-zero entries: {non_zero_entries:,}")
print(f"Total possible entries: {total_possible_entries:,}")
print(f"Sparsity: {sparsity:.8f}")

# Product-Product Co-occurrence Sparsity

TOP_K = 20

# Select top K most frequent products
top_products = (
    op_prior_m["product_id"]
    .value_counts()
    .head(TOP_K)
    .index
)

# Filter dataset to only those products
filtered = op_prior_m[op_prior_m["product_id"].isin(top_products)]

# Create basket-product pivot table (sparse)
basket_product = (
    filtered
    .assign(value=1)
    .pivot_table(index="order_id",
                 columns="product_id",
                 values="value",
                 fill_value=0)
)

# Compute co-occurrence matrix using matrix multiplication
co_matrix = basket_product.T.dot(basket_product)

# Remove diagonal (self-co-occurrence)
np.fill_diagonal(co_matrix.values, 0)

# Count non-zero co-occurrences
non_zero_pairs = (co_matrix.values > 0).sum()

# Total possible product-product pairs
total_pairs = TOP_K * (TOP_K - 1)

co_sparsity = 1 - (non_zero_pairs / total_pairs)

print("\n=== Product-Product Co-occurrence Sparsity (Top 100) ===")
print(f"Top K products: {TOP_K}")
print(f"Non-zero co-occurring pairs: {non_zero_pairs:,}")
print(f"Total possible pairs: {total_pairs:,}")
print(f"Co-occurrence Sparsity: {co_sparsity:.6f}")

# Tests
# Non-zero entries cannot exceed total possible entries
assert non_zero_entries <= total_possible_entries, \
    "ERROR: Non-zero entries exceed total possible entries!"
assert non_zero_pairs <= total_pairs, \
    "ERROR: Non-zero co-occurring pairs exceed total possible pairs!"

# Basket-product matrix dimensions match expectations
assert basket_product.shape[1] == TOP_K, \
    "ERROR: Basket-product matrix does not contain TOP_K columns!"
assert co_matrix.shape == (TOP_K, TOP_K), \
    "ERROR: Co-occurrence matrix dimensions incorrect!"
```

### (E) Initial Insights and Direction – Ex:

Observation: Most items appear in fewer than 1% of transactions
Hypothesis: High support thresholds miss meaningful temporal patterns

Potential RQs:
- How do different support thresholds affect rule quality?
- Do sequential patterns reveal structure missed by frequent itemsets?

1. Observation: Product demand is extremely skewed (long-tail distribution).
Only 1.58% of products account for 50% of all line-items, and fewer than 10% cover 80% of purchases. A very small group of products (e.g., bananas, organic bananas, strawberries) dominates total volume, while the majority of products contribute marginally. The hypothesis could be, traditional frequent itemset mining with moderate support thresholds will only capture popular staples and ignore niche but behaviorally meaningful items. So, instead of relying solely on high support thresholds, we should experiment with lower support thresholds combined with lift/confidence filtering. Another direction would be to mine patterns within departments (e.g., produce, dairy) separately to avoid dominance from high-frequency staple items.

2. Observation: Reordering behavior is strong and category-dependent.
The mean reorder ratio for non-zero baskets is ~0.60, indicating that most baskets consist largely of previously purchased items. Departments like dairy eggs (0.67), beverages (0.65), and produce (0.65) have especially high reorder rates, while pantry and personal care show lower repeat behavior. It could be because grocery shopping is driven heavily by habitual purchasing, particularly for perishable staples such as milk and fresh produce. Sequential pattern mining or temporal rule mining may be more meaningful than static association rules. Studying repeat cycles (e.g., milk every X days) could reveal stronger behavioral patterns than simple co-occurrence rules.

3. Observation: Time-of-day influences reorder intensity and basket size.
Orders peak between 9 AM and 3 PM, but reorder ratios are highest during early morning hours (6–8 AM). Late evening orders (21–23) show slightly larger basket sizes. This suggests that customer intent varies across time windows. Early morning purchases may represent routine replenishment trips, while evening orders may reflect larger planned grocery runs. So, Temporal segmentation should be incorporated into pattern mining. Association rules conditioned on hour-of-day or day-of-week may produce higher-quality and more interpretable patterns than global rules.

4. Observation: Popularity and reorder rate are not perfectly aligned.
Some highly popular products have relatively lower reorder rates (e.g., impulse or occasional-use items like fresh herbs or specialty ingredients), while moderately popular products (especially milk variants) show extremely high reorder rates (>0.85). Future analysis could cluster products based on both popularity and reorder rate to identify core staples vs occasional complements. This segmentation can guide rule generation, recommendation strategies, and personalized modeling.
